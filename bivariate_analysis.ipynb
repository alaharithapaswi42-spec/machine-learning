```python
# ============================================
# BIVARIATE ANALYSIS
# ============================================

# --------------------------------------------
# 1. IMPORT LIBRARIES
# --------------------------------------------

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from google.colab import files


# --------------------------------------------
# 2. UPLOAD DATASET
# --------------------------------------------

print("Please upload your CSV file:")

uploaded = files.upload()

# Automatically get the uploaded file name
file_name = list(uploaded.keys())[0]

# Read CSV file
df = pd.read_csv(file_name)

print("\nDataset loaded successfully!")


# --------------------------------------------
# 3. DISPLAY FIRST 5 ROWS
# --------------------------------------------

print("\nFirst 5 rows of the dataset:")
print(df.head())


# --------------------------------------------
# 4. DISPLAY COLUMN NAMES
# --------------------------------------------

print("\nColumn names:")
print(df.columns.tolist())


# --------------------------------------------
# 5. DATASET SHAPE
# --------------------------------------------

print("\nDataset shape:")
print(df.shape)


# --------------------------------------------
# 6. CHECK MISSING VALUES
# --------------------------------------------

print("\nMissing values:")
print(df.isnull().sum())


# ============================================
# 7. SCATTER PLOT
# CGPA vs Attendance Percentage
# ============================================

plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x="CGPA",
    y="AttendancePercent"
)

plt.title("CGPA vs Attendance Percentage")
plt.xlabel("CGPA")
plt.ylabel("Attendance Percentage")

plt.tight_layout()
plt.show()


# ============================================
# 8. REGRESSION PLOT
# CGPA vs Coding Test Score
# ============================================

plt.figure(figsize=(7, 5))

sns.regplot(
    data=df,
    x="CGPA",
    y="CodingTestScore"
)

plt.title("CGPA vs Coding Test Score")
plt.xlabel("CGPA")
plt.ylabel("Coding Test Score")

plt.tight_layout()
plt.show()


# ============================================
# 9. BOX PLOT
# CGPA vs Placement Status
# ============================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="PlacementStatus",
    y="CGPA"
)

plt.title("CGPA vs Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("CGPA")

plt.tight_layout()
plt.show()


# ============================================
# 10. VIOLIN PLOT
# Attendance vs Placement Status
# ============================================

plt.figure(figsize=(7, 5))

sns.violinplot(
    data=df,
    x="PlacementStatus",
    y="AttendancePercent"
)

plt.title("Attendance vs Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Attendance Percentage")

plt.tight_layout()
plt.show()


# ============================================
# 11. GENDER vs PLACEMENT STATUS
# ============================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Gender",
    hue="PlacementStatus"
)

plt.title("Gender vs Placement Status")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ============================================
# 12. STREAM vs PLACEMENT STATUS
# ============================================

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Stream",
    hue="PlacementStatus"
)

plt.title("Stream vs Placement Status")
plt.xlabel("Stream")
plt.ylabel("Number of Students")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================
# 13. CORRELATION
# ============================================

print("\nCorrelation Matrix:")

# Select only numerical columns
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

print(correlation)


# ============================================
# 14. CORRELATION HEATMAP
# ============================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()


# ============================================
# 15. BASIC STATISTICS
# ============================================

print("\nBasic Statistical Summary:")
print(df.describe())
```
